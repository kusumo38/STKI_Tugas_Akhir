# PRAKTIKUM BOW

```python
import pandas as pd

df = pd.read_csv('clean_dataset_stem.csv', sep=';', on_bad_lines='skip')

if df.shape[1] == 1:
    df = pd.read_csv('clean_dataset_stem.csv', sep=',', on_bad_lines='skip')

print("Daftar Kolom yang Ada di Dataset:")
print(df.columns.tolist())

print("\nUkuran Dataset Terbaru:", df.shape)
print("\nDistribusi Label Sentimen:")
print(df['Sentimen'].value_counts())
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# Menangani nilai NaN jika ada pada kolom teks
X_text = df['ProcessedText'].fillna('')

# 1. BOW Unigram (Kata Tunggal)
bow_unigram = CountVectorizer()
X_bow_uni = bow_unigram.fit_transform(X_text)

print("--- BAG OF WORDS (UNIGRAM) ---")
print("Jumlah Dokumen & Vocabulary Size:", X_bow_uni.shape)
print("Contoh 10 Kata Pertama di Vocabulary:", list(bow_unigram.vocabulary_.keys())[:10])

# 2. BOW Bigram (Pasangan Dua Kata)
bow_bigram = CountVectorizer(ngram_range=(2, 2))
X_bow_bi = bow_bigram.fit_transform(X_text)

print("\n--- BAG OF WORDS (BIGRAM) ---")
print("Jumlah Dokumen & Vocabulary Size:", X_bow_bi.shape)
print("Contoh 10 Pasangan Kata di Vocabulary:", list(bow_bigram.vocabulary_.keys())[:10])
```

# PRAKTIKUM TF-IDF Vectorizer

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Ekstraksi Fitur TF-IDF
tfidf_vec = TfidfVectorizer()
X_tfidf = tfidf_vec.fit_transform(X_text)

print("--- TF-IDF VECTORIZER ---")
print("Matriks TF-IDF Shape:", X_tfidf.shape)

# menampilkan nilai TF-IDF dalam bentuk DataFrame untuk 5 data pertama
df_tfidf_sample = pd.DataFrame(
    X_tfidf[:5].toarray(),
    columns=tfidf_vec.get_feature_names_out()
)
print("\nPreview Matriks TF-IDF (5 Dokumen Pertama):")
display(df_tfidf_sample.head())
```
