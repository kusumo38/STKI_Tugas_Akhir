# Studi Kasus: Sistem Manajemen & Kasir Toko Buku ("Toko Buku Cerdas")

# Variabel, Input/Output, & Tipe Data

```python
# --- Pendaftaran Toko & Input Pengguna ---
print("==========================================")
print("     SISTEM KASIR TOKO BUKU CERDAS        ")
print("==========================================")

# Mengambil input dari pengguna
nama_kasir = input("Masukkan nama kasir: ")
nama_pembeli = input("Masukkan nama pembeli: ")

# Variabel dan tipe data
total_belanja = 0
status_member = True  # Tipe data boolean

print(f"\nSelamat bertugas, {nama_kasir}!")
print(f"Melayani pembeli: {nama_pembeli} (Status Member: {status_member})")
```

# Operator, Pengkondisian, Iterasi, & List

```python
# --- Manajamen Inventaris (List) & Transaksi (Looping + If-Else) ---

# Daftar harga buku (List)
daftar_buku = ["Pemrograman Python", "Data Science Dasar", "Logika Algoritma"]
daftar_harga = [50000, 75000, 60000]

print("\n--- DAFTAR BUKU TERSEDIA ---")
for i in range(len(daftar_buku)):
    print(f"{i+1}. {daftar_buku[i]} - Rp{daftar_harga[i]}")

# Simulasi Pembelian
keranjang_harga = []
jumlah_pembelian = int(input("\nMasukkan jumlah jenis buku yang ingin dibeli: "))

for k in range(jumlah_pembelian):
    pilihan = int(input(f"Pilih nomor buku ke-{k+1} (1-3): ")) - 1
    jumlah_qty = int(input(f"Jumlah eksemplar '{daftar_buku[pilihan]}': "))

    subtotal = daftar_harga[pilihan] * jumlah_qty
    keranjang_harga.append(subtotal)

# Menghitung Total Belanja menggunakan Operator & Loop
total_belanja = sum(keranjang_harga)

# Pengkondisian Diskon (If-Else)
diskon = 0
if total_belanja >= 150000 and status_member:
    diskon = 0.20  # Diskon 20%
elif total_belanja >= 100000:
    diskon = 0.10  # Diskon 10%
else:
    diskon = 0.0

total_bayar = total_belanja - (total_belanja * diskon)

print(f"\nTotal Belanja kotor : Rp{total_belanja}")
print(f"Diskon              : {int(diskon * 100)}%")
print(f"Total Bayar Akhir   : Rp{total_bayar}")
```

# Fungsi, Tuple, & Dictionary

```python
# --- Struktur Data Dictionary & Fungsi ---

# Menggunakan Dictionary untuk menyimpan rincian buku
katalog_buku = {
    "B001": {"judul": "Pemrograman Python", "harga": 50000, "stok": 10},
    "B002": {"judul": "Data Science Dasar", "harga": 75000, "stok": 5},
    "B003": {"judul": "Logika Algoritma", "harga": 60000, "stok": 8}
}

# Fungsi untuk menghitung total pembayaran beserta cetak struk
def hitung_transaksi(kode_buku, jumlah, is_member=False):
    if kode_buku in katalog_buku:
        buku = katalog_buku[kode_buku]
        if buku["stok"] >= jumlah:
            subtotal = buku["harga"] * jumlah
            buku["stok"] -= jumlah  # Mengurangi stok

            # Hitung diskon member
            potongan = subtotal * 0.15 if is_member else 0
            total = subtotal - potongan

            return {
                "judul": buku["judul"],
                "subtotal": subtotal,
                "potongan": potongan,
                "total": total,
                "sisa_stok": buku["stok"]
            }
        else:
            return "Stok tidak mencukupi!"
    else:
        return "Kode buku tidak ditemukan!"

# Memanggil Fungsi
hasil_transaksi = hitung_transaksi("B002", 2, is_member=True)
print("\n--- HASIL TRANSAKSI FUNGSI ---")
print(hasil_transaksi)
```

# Modul & Packages

```python
# --- Menggunakan Modul Bawaan Python ---
import datetime
import math

waktu_transaksi = datetime.datetime.now()
pembayaran_pembeli = 150000
kembalian = pembayaran_pembeli - hasil_transaksi["total"]

print("\n================ STRUK PEMBAYARAN ================")
print(f"Tanggal Transaksi : {waktu_transaksi.strftime('%d-%m-%Y %H:%M:%S')}")
print(f"Item Dibeli       : {hasil_transaksi['judul']}")
print(f"Total Tagihan     : Rp{math.ceil(hasil_transaksi['total'])}") # Menggunakan math.ceil
print(f"Tunai             : Rp{pembayaran_pembeli}")
print(f"Kembalian         : Rp{kembalian}")
print("==================================================")
```

# Pemrograman Berbasis Objek / OOP

```python

# --- Menerapkan Class dan Object (OOP) ---

class Buku:
    def __init__(self, kode, judul, harga, stok):
        self.kode = kode
        self.judul = judul
        self.harga = harga
        self.stok = stok

    def kurangi_stok(self, jumlah):
        if self.stok >= jumlah:
            self.stok -= jumlah
            return True
        return False

    def info_buku(self):
        return f"[{self.kode}] {self.judul} | Harga: Rp{self.harga} | Sisa Stok: {self.stok}"


class TransaksiKasir:
    def __init__(self, nama_kasir):
        self.nama_kasir = nama_kasir
        self.daftar_item = []

    def tambah_item(self, buku, jumlah):
        if buku.kurangi_stok(jumlah):
            self.daftar_item.append((buku, jumlah))
            print(f"✅ Berhasil menambahkan {jumlah} x '{buku.judul}' ke keranjang.")
        else:
            print(f"❌ Gagal! Stok '{buku.judul}' tidak cukup.")

    def cetak_ringkasan(self):
        print(f"\n--- RINGKASAN TRANSAKSI (Kasir: {self.nama_kasir}) ---")
        total = 0
        for buku, qty in self.daftar_item:
            subtotal = buku.harga * qty
            total += subtotal
            print(f"- {buku.judul} ({qty}x) = Rp{subtotal}")
        print(f"TOTAL AKHIR: Rp{total}")


# --- Penggunaan Class & Object ---
# 1. Membuat Instansiasi Objek Buku
buku1 = Buku("B01", "Python untuk Pemula", 50000, 10)
buku2 = Buku("B02", "Algoritma & Struktur Data", 80000, 3)

# 2. Membuat Objek Kasir
kasir_hari_ini = TransaksiKasir("Budi")

# 3. Menjalankan Method
kasir_hari_ini.tambah_item(buku1, 2)
kasir_hari_ini.tambah_item(buku2, 5)  # Harusnya gagal karena stok hanya 3
kasir_hari_ini.tambah_item(buku2, 2)  # Berhasil

# 4. Cetak Ringkasan Transaksi
kasir_hari_ini.cetak_ringkasan()

# 5. Cek Sisa Stok
print("\n--- SISA STOK TERKINI ---")
print(buku1.info_buku())
print(buku2.info_buku())
```
